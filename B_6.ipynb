```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# 1. Read dataset
df = pd.read_csv("Assignment-1_Data.csv",
                 encoding="unicode_escape", low_memory=False)

# 2. Clean data
df.columns = df.columns.str.replace("ï»¿", "", regex=False).str.strip()
df = df.dropna(subset=["CustomerID"])
df["BillNo"] = df["BillNo"].astype(str)
df = df[~df["BillNo"].str.contains("C")]
df["Itemname"] = df["Itemname"].astype(str).str.strip()

# 3. Create transactions
transactions = df.groupby("BillNo")["Itemname"].apply(list).tolist()
print("Total transactions:", len(transactions))

# 4. Convert into encoded format
te = TransactionEncoder()
data = te.fit(transactions).transform(transactions)
encoded = pd.DataFrame(data, columns=te.columns_)

# 5. Find frequent itemsets
frequent = apriori(encoded, min_support=0.002, use_colnames=True)
print(frequent)

# 6. Generate association rules
rules = association_rules(frequent, metric="confidence",
                          min_threshold=0.5)
print(rules)
```
